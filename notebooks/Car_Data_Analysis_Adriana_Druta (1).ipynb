###### **06.1.1 Group by Driven Wheels** for columns MSRP and Popularity / (calculate the average)

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Gruppieren nach Antriebsart
df_grouped = df_cleaned.groupby(['driven_wheels'])[['msrp', 'popularity']].mean().round(2)

# 2. Visualisierung
df_grouped.plot(kind='bar', subplots=True, layout=(1, 2), figsize=(12, 5))
farbe_msrp = 'midnightblue'
farbe_pop = 'darkorange'
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for readability

# 3. Titel & Layout
plt.suptitle('Mean MSRP and Popularity by Driven Wheels')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
df_grouped = df_cleaned.groupby(['driven_wheels'])[['msrp', 'popularity']].mean().round(2)

fig, ax1 = plt.subplots(figsize=(14, 10))

farbe_msrp = 'midnightblue'
df_grouped['msrp'].plot(kind='bar', ax=ax1, color=farbe_msrp, alpha=0.7, label='msrp', width=0.4, position=1)
ax1.set_ylabel('Mean MSRP ($)', color=farbe_msrp)
ax1.tick_params(axis='y', labelcolor=farbe_msrp)
ax1.set_xlabel('Driven Wheels')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for readability
ax2 = ax1.twinx()
farbe_pop = 'darkorange'
df_grouped['popularity'].plot(kind='bar', ax=ax2, color=farbe_pop, alpha=0.7, label='Popularity', width=0.4, position=0)
ax2.set_ylabel('Mean Popularity', color=farbe_pop)
ax2.tick_params(axis='y', labelcolor=farbe_pop)

plt.title('Mean MSRP and Popularity by Driven Wheels', pad=15)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', bbox_to_anchor=(0.6, 0.95))

plt.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(16, 8))

df_grouped = df_cleaned.groupby(['engine_cylinders']).agg({'msrp': 'mean', 'popularity': 'mean'}).round(2)

categories = df_grouped.index
x = range(len(categories))
width = 0.35

# Plot MSRP on left y-axis
bars1 = ax1.bar([i - width/2 for i in x], df_grouped.msrp, width, label='msrp', color='steelblue')
ax1.set_xlabel('Engine_Cylinders')
ax1.set_ylabel('Mean MSRP', color='midnightblue')
ax1.tick_params(axis='y', labelcolor='midnightblue')

# Plot Popularity on right y-axis
ax2 = ax1.twinx()
bars2 = ax2.bar([i + width/2 for i in x], df_grouped.popularity, width, label='popularity', color='darkorange')
ax2.set_ylabel('Mean Popularity', color='darkorange')
ax2.tick_params(axis='y', labelcolor='darkorange')

# X-axis labels
ax1.set_xticks(list(x))
ax1.set_xticklabels(categories, rotation=45, ha='right')

# Combined legend
lines = [bars1, bars2]
ax1.legend(lines, ['MSRP', 'Popularity'], loc='upper left')

plt.title('Mean MSRP and Popularity by Engine Cylinders')
plt.tight_layout()
plt.show()
```

#####**ouliers analysis**

```python
midsize_df = df_cleaned[df_cleaned['vehicle_size'] == 'Midsize']

def find_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    return outliers, lower_bound, upper_bound

# Check MSRP outliers for Midsize vehicles
msrp_outliers, msrp_lower, msrp_upper = find_outliers_iqr(midsize_df['msrp'])

# Check Engine Cylinders outliers for Midsize vehicles
engine_cyl_outliers, engine_cyl_lower, engine_cyl_upper = find_outliers_iqr(midsize_df['engine_cylinders'])

print(f"\n--- MSRP Outliers for Midsize Vehicles ---")
print(f"Q1: {midsize_df['msrp'].quantile(0.25):.2f}, Q3: {midsize_df['msrp'].quantile(0.75):.2f}, IQR: {midsize_df['msrp'].quantile(0.75) - midsize_df['msrp'].quantile(0.25):.2f}")
print(f"Lower Bound (MSRP): {msrp_lower:.2f}, Upper Bound (MSRP): {msrp_upper:.2f}")
print(f"Number of MSRP outliers: {len(msrp_outliers)}")
if not msrp_outliers.empty:
    print(f"Min MSRP outlier: {msrp_outliers.min():.2f}, Max MSRP outlier: {msrp_outliers.max():.2f}")

print(f"\n--- Engine Cylinders Outliers for Midsize Vehicles ---")
print(f"Q1: {midsize_df['engine_cylinders'].quantile(0.25):.2f}, Q3: {midsize_df['engine_cylinders'].quantile(0.75):.2f}, IQR: {midsize_df['engine_cylinders'].quantile(0.75) - midsize_df['engine_cylinders'].quantile(0.25):.2f}")
print(f"Lower Bound (Engine Cylinders): {engine_cyl_lower:.2f}, Upper Bound (Engine Cylinders): {engine_cyl_upper:.2f}")
print(f"Number of Engine Cylinders outliers: {len(engine_cyl_outliers)}")
if not engine_cyl_outliers.empty:
    print(f"Engine Cylinders Outliers:\n{engine_cyl_outliers.value_counts().sort_index()}")
```
